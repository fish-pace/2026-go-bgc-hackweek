## Classic approaches for estimating subsurface and integrated chlorophyll from surface Chl-a

**Straight from ChatGPT ...**

A common satellite-ocean-color problem is to infer the vertical chlorophyll distribution from surface chlorophyll-a (`Chl_surface`). Classic empirical approaches use surface Chl-a as an indicator of trophic state and predict quantities such as:

- euphotic depth (`Z_e`). Note our training dataset has Kd_490 which can get at `Z_e`
- the vertical chlorophyll profile, `Chl(z)`
- depth and magnitude of the deep chlorophyll maximum (DCM)
- vertically integrated chlorophyll over the euphotic zone

### Morel & Berthon (1989)

Morel and Berthon (1989) developed empirical relationships between near-surface chlorophyll, euphotic depth, and vertically integrated pigment concentration.

A commonly used relationship between euphotic depth and integrated chlorophyll is approximately:

$$C_{\mathrm{tot}} = 4910\,Z_e^{-1.34}$$

where:

- \(C_{\mathrm{tot}}\) is vertically integrated chlorophyll/pigment concentration in mg m\(^{-2}\)
- \(Z_e\) is the euphotic depth in meters

They also described normalized vertical chlorophyll profiles whose shape varies with trophic state. Conceptually,

$$
Chl(z) = Chl_{\mathrm{surface}}\,
f\left(\frac{z}{Z_e}; Chl_{\mathrm{surface}}\right)
$$

where \(f\) represents an empirically derived vertical-profile shape.

In oligotrophic waters with low surface Chl-a, the predicted profile generally includes a pronounced subsurface chlorophyll maximum. In more productive waters, chlorophyll tends to be more uniformly distributed through the upper water column or concentrated near the surface.

### Uitz et al. (2006)

Uitz et al. (2006) revisited the Morel & Berthon approach using a much larger collection of HPLC chlorophyll profiles.

They parameterized normalized vertical chlorophyll profiles as a function of trophic state derived from surface Chl-a. Their formulation provides a way to estimate:

$$
Chl_{\mathrm{surface}}
\rightarrow
Chl(z)
$$

and therefore quantities such as:

$$
z_{\mathrm{DCM}}, \qquad
Chl_{\mathrm{DCM}}, \qquad
\int_0^{Z_e} Chl(z)\,dz
$$

The Uitz et al. formulation has been widely used in satellite-based estimates of phytoplankton vertical structure and primary production.

### Conceptual satellite-to-profile workflow

The classic empirical approach can be summarized as:

$$
Chl_{\mathrm{surface}}
\rightarrow
\begin{cases}
Z_e \\
Chl(z) \\
z_{\mathrm{DCM}} \\
Chl_{\mathrm{DCM}} \\
\displaystyle \int_0^{Z_e} Chl(z)\,dz
\end{cases}
$$

These relationships provide useful baseline models for evaluating a 3-D chlorophyll product such as CHLA-Z.

### Important caveat: DCM versus biomass maximum

A deep chlorophyll maximum (DCM) is not necessarily a deep phytoplankton biomass maximum.

At depth, phytoplankton can increase their cellular chlorophyll content in response to lower light levels (photoacclimation). Consequently, chlorophyll concentration can increase without a proportional increase in phytoplankton carbon or biomass.

This distinction is important when interpreting reconstructed vertical chlorophyll profiles.

### More recent BGC-Argo work

Modern BGC-Argo profile datasets make it possible to revisit these classic surface-to-column relationships using global in situ observations.

For CHLA-Z, useful baseline comparisons would therefore include:

1. Morel & Berthon (1989) surface-Chl / integrated-Chl relationships
2. Morel & Berthon normalized vertical profiles
3. Uitz et al. (2006) vertical-profile parameterization
4. BGC-Argo-based revisions or evaluations of surface-to-integrated and surface-to-DCM relationships: [Tanner et al. (2024)](https://www.frontiersin.org/journals/remote-sensing/articles/10.3389/frsen.2024.1495958/full) "Revisiting the relationship between surface and column-integrated chlorophyll-a concentrations in the Biogeochemical-Argo and satellite era"
